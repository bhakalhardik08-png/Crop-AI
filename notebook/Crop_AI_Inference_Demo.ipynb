# 🌾 Crop AI — Inference Demo

This notebook demonstrates the released two-stage Crop AI pipeline using the trained champion models.

**Crop classes:** Banana, Guava, Maize, Rice, Wheat  
**Guava quality classes:** A, B, C, Reject

```python
!pip install -q tensorflow pillow numpy

import numpy as np
import tensorflow as tf
from tensorflow.keras.utils import load_img, img_to_array

CROP_MODEL_PATH = 'model/CROP_MODEL_CHAMPION_91_76_TEST.keras'
QUALITY_MODEL_PATH = 'model/CROP_QUALITY_MODEL_CHAMPION_78_27_TEST.keras'
CROP_CLASSES = ['Banana', 'Guava', 'Maize', 'Rice', 'Wheat']
QUALITY_CLASSES = ['A', 'B', 'C', 'Reject']

crop_model = tf.keras.models.load_model(CROP_MODEL_PATH)
quality_model = tf.keras.models.load_model(QUALITY_MODEL_PATH)
print('Models loaded successfully.')
```

```python
IMAGE_PATH = 'path/to/your/image.jpg'  # change this

img = load_img(IMAGE_PATH, target_size=(224, 224))
arr = np.expand_dims(img_to_array(img).astype('float32'), axis=0)

crop_probs = crop_model.predict(arr, verbose=0)[0]
crop_idx = int(np.argmax(crop_probs))
crop = CROP_CLASSES[crop_idx]
crop_conf = float(crop_probs[crop_idx] * 100)

print(f'Crop: {crop}')
print(f'Crop confidence: {crop_conf:.2f}%')

if crop == 'Guava' and crop_conf >= 50:
    quality_probs = quality_model.predict(arr, verbose=0)[0]
    quality_idx = int(np.argmax(quality_probs))
    print(f'Guava quality: {QUALITY_CLASSES[quality_idx]}')
    print(f'Quality confidence: {quality_probs[quality_idx] * 100:.2f}%')
else:
    print('Quality grading: not available for this prediction')
```
